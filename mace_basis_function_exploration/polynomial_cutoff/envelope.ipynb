```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def fenv(d, p):
    return (1 - ((p + 1)*(p + 2))/(2)* d**p + p*(p + 2)*d**(p+1) - (p*(p + 1))/(2) * d**(p+2))

```

```python
ds = np.linspace(0, 1, 100)
ps = np.arange(0, 10)

plt.subplots(1, 1, figsize = (8, 4))

for p in ps:
    ys = np.array([
        fenv(d, p)
        for d in ds
    ])
    plt.plot(ds, ys, label = f"{p}")

plt.legend(title="p", ncol = 2, loc = (1,0.1))
plt.xlabel("d (adimensional)")
plt.ylabel(r"$f_{\text{env}}$ (adimensional)")

plt.tight_layout()
plt.savefig("envelope_functions.pdf")
```
