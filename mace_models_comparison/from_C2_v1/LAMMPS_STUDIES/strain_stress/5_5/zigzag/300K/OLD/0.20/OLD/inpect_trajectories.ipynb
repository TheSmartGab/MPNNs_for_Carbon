```python
from ase.io import read
import numpy as np
```

```python
traj1 = read("dump_new.dump", index=":")
traj2 = read("dump_v3.dump", index="0:11")

print(len(traj1))
print(len(traj2))
```

```python
def compare_frames(f1, f2):
    if np.allclose(f1.get_positions()-f2.get_positions(), 0):
        return True
    
    return False
```

```python
for f1, f2 in zip(traj1, traj2):
    print(compare_frames(f1, f2))
```
