```python
from ase.visualize import view
from ase.build import graphene

import numpy as np

from mace.calculators import MACECalculator

from ase.optimize import BFGS
```

```python
calc = MACECalculator("../../MACE.model", device="cuda")

atoms = graphene()
atoms.calc = calc

opt = BFGS(atoms)
opt.run(fmax=1e-3)
view(atoms, viewer="x3d")
```

```python
cell0 = atoms.cell.copy()
cell = atoms.cell
```

```python
atoms.cell
```

```python
print(cell0)
```

```python
strain = np.array([
    [0.3, 0,0],
    [0,0.,0],
    [0,0,0]
])

atoms.set_cell((np.eye(3) + strain) @ cell0, scale_atoms=True)
atoms.calc = calc

opt = BFGS(atoms)
opt.run(fmax=1e-3)

stress = atoms.get_stress(voigt=False)

print(atoms.cell)
print(np.linalg.norm(atoms.cell, axis = -1))
print(stress)
print(np.trace(stress.T @ stress))
view(atoms, viewer="x3d")
```
